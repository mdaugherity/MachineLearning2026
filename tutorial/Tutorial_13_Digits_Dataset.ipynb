```python
import numpy as np
import matplotlib.pyplot as plt

# Common utilities
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import ConfusionMatrixDisplay

# Classifiers
from sklearn.naive_bayes import GaussianNB
```

# Tutorial 13: The Digits Dataset
The digits dataset in sklearn is a sample of a [more complete dataset](http://archive.ics.uci.edu/ml/datasets/Pen-Based+Recognition+of+Handwritten+Digits).  Ours has 1,797 examples of classified hand-written digits 0-9 stored as pixel arrays with color from 0 (white) to 15 (black).



# Load Data

```python
X,y = load_digits(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=0) # stratify ensures an equal ratio of each target
print("Data:  ",X.shape,y.shape)
print("Train: ",X_train.shape, y_train.shape)
print("Test:  ",X_test.shape,y_test.shape)
```

Let's try training a classifier and seeing if it works

# Classifier Performance
## The Training Data

```python
clf = GaussianNB()  # No parameters to pick!
clf.fit(X_train,y_train)  # only train on the training data!!!!!!!!!!!
clf.score(X_train,y_train)
```

The gaussian fits get 84.8% of the training data right.  Where did it go wrong?


As always, the only fair way to evaluate a classifier is to use data **it hasn't been trained on**. How well do we do with new samples?

```python
clf.score(X_test,y_test)
```

We get 83.6% right!  Consistent scores between the test and train dataset are a good sign that we are not under or over fitting.

```python
y_pred = clf.predict(X_test)
ConfusionMatrixDisplay.from_predictions(y_test, y_pred,cmap='Blues')
plt.show()
```

We get a nice visualization of our mistakes. We often confuse 2's as 8's. So if I were going to try to improve performance then those are the places I would start.

Let's examine the missed samples

```python
# Find the miss-classified samples
miss = []  # save missed sample numbers
#print('Sample\tTruth\tPredicted')
for i,pred in enumerate(y_pred):
    if pred!=y_test[i]:
        #print(f"{i}\t{y_test[i]}\t{pred}")
        miss.append(i)
```

```python
# Plot the first 5 missed samples
fig = plt.figure(figsize=(15, 3))
#fig.subplots_adjust(left=0, right=1, bottom=0, top=1, hspace=0.05, wspace=0.05) # make small panels

for i in range(5):
    m = miss[i]  # the sample number we are looking at
    ax = fig.add_subplot(1,5, i + 1, xticks=[], yticks=[])
    ax.imshow(X_test[m].reshape(8,8), cmap=plt.cm.binary)
    plt.title(f"Truth={y_test[m]}, Pred={y_pred[m]}")

plt.show()
```

Some of these are reasonable guesses, but some are pretty strange.

# WARNING - Don't Use Standard Scaler!

**Note:** I noticed that my performance seems a smidge worse if it is scaled.  This is an artifact of processing.

First, the problem:

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
```

```python
estimators = [('scaler', StandardScaler()),('clf', KNeighborsClassifier()) ] # see note at end
pipe = Pipeline(estimators)
pipe.fit(X_train, y_train)
print('Train score\t', pipe.score(X_train,y_train))
print('Test score\t', pipe.score(X_test,y_test))
```

```python
clf = KNeighborsClassifier()
clf.fit(X_train, y_train)
print('Train score\t', clf.score(X_train,y_train))
print('Test score\t', clf.score(X_test,y_test))
```

The pixels on the left and right columns are usually zero, except for the occasional edge.  This really skews the mean and variance, so StandardScaler badly distorts some pixels

```python
s = StandardScaler()
s.fit(X_train)
```

```python
plt.figure()
plt.subplot(1,3,1)
plt.imshow(X.max(axis=0).reshape(8,8), cmap='binary')
plt.title('Max')

plt.subplot(1,3,2)
plt.imshow(s.mean_.reshape(8,8), cmap='binary')
plt.title('Mean')

plt.subplot(1,3,3)
plt.imshow(s.var_.reshape(8,8), cmap='Reds')
plt.title('Var')

plt.show()
```
